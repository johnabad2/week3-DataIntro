## 2. Read Basic Python Syntax

Python uses readable structure. `#` starts a comment, `=` assigns a value to a name, parentheses call a function, and indentation groups statements. Names usually use `snake_case`; they cannot start with a number or contain spaces.

```python
# A comment explains the code but is ignored when Python runs the cell.
student_name = "Maya"  # Assignment gives a value a name.
score = 8

# The condition is checked first. Only one indented block will run.
if score >= 5:
    result = "pass"
else:
    result = "try again"

# print() displays several values separated by spaces.
print(student_name, "scored", score, "and will", result)
```

## 3. Work with Variables and Core Data Types

A variable is a name pointing to a value. Python figures out the type at runtime. The most common beginner types are integers (`int`), decimal numbers (`float`), text (`str`), true/false values (`bool`), and the special absence value `None`.

```python
# Variables give useful names to values.
age = 21
height_m = 1.72
name = "Aisha"
is_student = True

# print() displays the values stored in the variables.
print(age)
print(height_m)
print(name)
print(is_student)

# Convert text to a number before doing arithmetic with it.
year_text = "2026"
year_number = int(year_text)
print(year_number + 1)
print("Python" + " " + "basics")
```

## 4. Use Operators and Expressions

An expression combines values and operators to produce a result. Arithmetic operators include `+`, `-`, `*`, `/`, `//` (whole-number division), `%` (remainder), and `**` (power). Comparisons produce booleans. Logical operators combine conditions, and `in` checks membership.

```python
items = 17
people = 5

# Arithmetic operators calculate new values.
print("remainder:", items % people)
print("average:", items / people)
print("power:", 2 ** 3)

# Comparisons create True or False values.
temperature = 18
is_mild = 15 <= temperature <= 22

# The 'in' operator checks whether a value appears in a collection.
has_coat = "coat" in ["coat", "hat"]

# 'and' is True only when both conditions are True.
comfortable = is_mild and has_coat
print(is_mild, has_coat, comfortable)
```

## 5. Control Flow with `if`, `for`, and `while`

Control flow decides which instructions run and how often. An `if` statement chooses a branch; a `for` loop visits each item; a `while` loop repeats while a condition stays true. `break` exits a loop, while `continue` skips to the next iteration.

```python
scores = [4, 7, 10, 3, 8]

# The for loop visits each score in the list.
for score in scores:
    if score < 5:
        continue  # Skip scores below the passing mark.
    if score == 10:
        print("Perfect score")
        break  # Stop once the perfect score is found.
    print("Passing score:", score)

# A while loop repeats while its condition is True.
attempts = 0
while attempts < 3:
    print("Attempt", attempts + 1)
    attempts += 1
```

## 6. Write and Call Functions

A function packages a task under a name. Parameters are inputs, and `return` sends a result back to the caller. A default argument is used when the caller does not provide that parameter.

```python
def percentage(part, whole=100):
    """Return part as a percentage of whole."""
    # return sends the calculated value back to the caller.
    return part / whole * 100

# The first call uses the default whole of 100.
print(percentage(25))

# The second call supplies both arguments.
print(percentage(2, 8))
```

## 7. Use Built-in Collections

Collections group values. Lists are ordered and changeable, tuples are ordered and usually treated as fixed, dictionaries map keys to values, and sets keep unique values. Python uses zero-based indexing, so the first item is at index `0`; a slice such as `items[1:3]` stops before index `3`.

```python
# A list stores values in order and can be changed.
fruit = ["apple", "banana", "cherry"]
print(fruit)

# Index 0 means the first item. A slice selects part of the list.
print(fruit[0])
print(fruit[1:])

# append() adds an item at the end; assignment changes an existing item.
fruit.append("date")
fruit[1] = "blueberry"
print(fruit)
```

A **tuple** is also ordered, but it is normally treated as fixed. A **dictionary** stores values with named keys.

```python
# A tuple keeps its values in order.
coordinates = (53.35, -6.26)
print(coordinates[0])

# A dictionary uses keys to look up values.
student = {"name": "Maya", "score": 8}
print(student["name"])

# Add a new key and value to the dictionary.
student["passed"] = student["score"] >= 5
print(student)
```

A **set** stores unique values, so repeated values appear only once.

```python
unique_scores = set([8, 8, 6, 10, 6])
print(unique_scores)
```

Dictionaries can be looped over. `.items()` gives one key and its value on each pass through the loop.

```python
for key, value in student.items():
    print(key, "->", value)
```

## 8. Import Modules and Use Libraries

A **module** is a Python file containing reusable code. A **library** is a collection of modules and tools. Importing makes that code available in the current notebook; it does not copy the code into every cell.

Use `import module`, `import module as short_name`, or `from module import name`. Later, you will import third-party libraries such as pandas; here you use only Python's standard library.

```python
# Import the math module and give statistics a shorter alias.
import math
import statistics as stats
from datetime import date

radius = 3
print("circle area:", math.pi * radius ** 2)
print("mean:", stats.mean([4, 7, 9]))
print("today's year:", date.today().year)

# The prefix shows where each imported name comes from.
print(math.sqrt(16))
```

## 9. Inspect Library Objects with `help()` and `dir()`

When you meet an unfamiliar module or function, `dir()` lists available names and `help()` shows documentation. In a notebook, the shorter `object?` form also opens help in VS Code/Jupyter. Reading a function's docstring is often the fastest way to learn its inputs and output.

```python
# dir() lists names available in a module.
public_math_names = [name for name in dir(math) if not name.startswith("_")]
print(public_math_names[:12])

# help() displays documentation for a function.
help(math.sqrt)

# A function's docstring records what it is intended to do.
print("Function documentation:")
print(percentage.__doc__)
```

## 10. Handle Common Errors and Exceptions

An error message is useful information: it names the error type and usually points to where Python stopped. `NameError` means a name is unavailable, while `TypeError` often means an operation received the wrong kind of value. Use `try` and `except` when you can handle an expected problem.

```python
def divide_numbers(numerator, denominator):
    try:
        return numerator / denominator
    except ZeroDivisionError:
        return "Cannot divide by zero"

print(divide_numbers(10, 2))
print(divide_numbers(10, 0))

# Uncomment one line at a time to see the error type and message.
# print(not_defined_yet)       # NameError: this name does not exist.
# print("score" + 8)            # TypeError: these types cannot be combined.
```
